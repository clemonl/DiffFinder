```python
from difffinder import compare

def count_to(n):            # your version
    total = 0
    for i in range(n):
        total += i
    return total

def count_to_ref(n):        # the reference
    total = 0
    for i in range(n + 1):
        total += i
    return total

compare(count_to, count_to_ref, args=(4,))
```

```python
from difffinder import compare_traces

def count_to(n):            # your version
    total = 0
    for i in range(n):
        total += i
    return total

def count_to_ref(n):        # the reference
    total = 0
    for i in range(n + 1):
        total += i
    return total

compare_traces(count_to, count_to_ref, args=(4,))
```

```python
from difffinder import compare_traces

def trays_needed(cups, capacity):
    cups = int(cups)
    capacity = int(capacity)
    if cups%capacity == 0:
        return int(cups//capacity)
    else: return int(cups//capacity + cups%capacity)

def trays_needed_r(cups, capacity): 
    return (cups + capacity - 1) // capacity

compare_traces(trays_needed, trays_needed_r, args = (1,1000))
```
